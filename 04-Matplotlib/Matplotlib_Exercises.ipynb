## Section 2 — Binning & Discretization

**Q6.** Cut `Age` into 4 equal-width bins with labels `['Young','Adult','MiddleAged','Senior']`. Store in `Age_Group`.

*Hint: `pd.cut(col, bins=4, labels=[...])`*

```python
# Q6 - Cut Age into 4 equal-width bins
df['Age_Group'] = pd.cut(df['Age'], bins=4, labels=['Young', 'Adult', 'MiddleAged', 'Senior'])
df['Age_Group'].value_counts()
```

**Q7.** Count how many rows fall in each `Age_Group`.

*Hint: `.value_counts()`*

```python
# Q7 - Count rows in each Age_Group
df['Age_Group'].value_counts()
```

**Q8.** Use `pd.qcut` to divide `Glucose` into 4 quantile bins with labels `['Low','Medium','High','VeryHigh']`. Store in `Glucose_Quantile`.

*Hint: `pd.qcut(col, q=4, labels=[...])`*

```python
# Q8 - qcut Glucose into 4 quantile bins
df['Glucose_Quantile'] = pd.qcut(df['Glucose'], q=4, labels=['Low', 'Medium', 'High', 'VeryHigh'])
df['Glucose_Quantile'].value_counts()
```

**Q9.** Use custom bin edges `[0, 60, 80, 90, 200]` to cut `BloodPressure` into categories `['Low','Normal','High','Crisis']`.

*Hint: `pd.cut(col, bins=[...], labels=[...])`*

```python
# Q9 - Custom bin edges for BloodPressure
df['BP_Category'] = pd.cut(df['BloodPressure'], bins=[0, 60, 80, 90, 200],
                            labels=['Low', 'Normal', 'High', 'Crisis'])
df['BP_Category'].value_counts()
```

## Section 4 — Matplotlib Basics

**Q14.** Create a **line plot** of the first 50 values of `Glucose`. Add title, xlabel, ylabel.

*Hint: `plt.plot(df['Glucose'].values[:50])`*

```python
# Q14 - Line plot of first 50 values of Glucose
plt.figure(figsize=(10, 4))
plt.plot(df['Glucose'].values[:50], color='steelblue', linewidth=1.5)
plt.title('Glucose - First 50 Values')
plt.xlabel('Index')
plt.ylabel('Glucose')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Q15.** Create a **scatter plot** of `Age` (x) vs `Glucose` (y). Color points by `Outcome` — green for 0, red for 1.

*Hint: `df['Outcome'].map({0:'green', 1:'red'})`*

```python
# Q15 - Scatter plot: Age vs Glucose, colored by Outcome
colors = df['Outcome'].map({0: 'green', 1: 'red'})
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Glucose'], c=colors, alpha=0.5, edgecolors='k', linewidths=0.3)
plt.title('Age vs Glucose (Green=No Diabetes, Red=Diabetes)')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.tight_layout()
plt.show()
```

**Q16.** Create a **histogram** of `BMI` with 20 bins. Add a vertical line at the mean.

*Hint: `plt.hist()` then `plt.axvline(df['BMI'].mean())`*

```python
# Q16 - Histogram of BMI with vertical mean line
plt.figure(figsize=(8, 5))
plt.hist(df['BMI'], bins=20, color='skyblue', edgecolor='white')
plt.axvline(df['BMI'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean = {df["BMI"].mean():.1f}')
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

**Q17.** Create **two overlapping histograms** — `Glucose` for Outcome=0 and Outcome=1 on the same axes. Use `alpha=0.6`. Add legend.

*Hint: call `plt.hist()` twice, one per group*

```python
# Q17 - Two overlapping histograms for Glucose by Outcome
plt.figure(figsize=(8, 5))
plt.hist(df[df['Outcome']==0]['Glucose'], bins=20, alpha=0.6, color='green', label='No Diabetes (0)')
plt.hist(df[df['Outcome']==1]['Glucose'], bins=20, alpha=0.6, color='red',   label='Diabetes (1)')
plt.title('Glucose Distribution by Outcome')
plt.xlabel('Glucose')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

**Q18.** Create a **horizontal bar chart** showing the mean of each feature (drop `Outcome`).

*Hint: `plt.barh(labels, values)`*

```python
# Q18 - Horizontal bar chart of mean feature values
means = df.drop(columns='Outcome').mean()
plt.figure(figsize=(8, 6))
plt.barh(means.index, means.values, color='steelblue', edgecolor='white')
plt.title('Mean Value of Each Feature')
plt.xlabel('Mean Value')
plt.tight_layout()
plt.show()
```

**Q19.** Create a **box plot** comparing `BMI` for diabetic vs non-diabetic patients.

*Hint: `plt.boxplot([group0_values, group1_values])`*

```python
# Q19 - Box plot: BMI for diabetic vs non-diabetic
group0 = df[df['Outcome']==0]['BMI'].values
group1 = df[df['Outcome']==1]['BMI'].values
plt.figure(figsize=(7, 5))
plt.boxplot([group0, group1], labels=['No Diabetes (0)', 'Diabetes (1)'], patch_artist=True,
            boxprops=dict(facecolor='lightblue'))
plt.title('BMI Distribution by Outcome')
plt.ylabel('BMI')
plt.tight_layout()
plt.show()
```

**Q20.** Create a **2×2 grid** of subplots showing histograms of `Glucose`, `BMI`, `Age`, and `BloodPressure`.

*Hint: `fig, axes = plt.subplots(2, 2, figsize=(12,8))`*

```python
# Q20 - 2x2 grid of histograms
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
cols = ['Glucose', 'BMI', 'Age', 'BloodPressure']
colors = ['steelblue', 'salmon', 'mediumseagreen', 'mediumpurple']
for ax, col, color in zip(axes.flat, cols, colors):
    ax.hist(df[col], bins=20, color=color, edgecolor='white')
    ax.set_title(f'{col} Distribution')
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
plt.suptitle('Feature Distributions', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

**Q21.** Save any one of your plots to a file called `my_plot.png`.

*Hint: `plt.savefig('my_plot.png', dpi=100, bbox_inches='tight')`*

```python
# Q21 - Save a plot to my_plot.png
plt.figure(figsize=(8, 5))
plt.hist(df['Glucose'], bins=20, color='steelblue', edgecolor='white')
plt.title('Glucose Distribution')
plt.xlabel('Glucose')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('my_plot.png', dpi=100, bbox_inches='tight')
plt.show()
print("Plot saved as my_plot.png")
```

## Bonus Challenge

**Q39 (Bonus).** In a single figure with `1 row × 3 columns`:
- Plot 1: KDE of Glucose for both Outcome groups
- Plot 2: Boxplot of BMI by Outcome (seaborn)
- Plot 3: Scatter of Age vs Glucose colored by Outcome

Add a title to each subplot. Use `figsize=(18, 5)`.

```python
# Q39 Bonus - 1x3 subplot figure
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: KDE of Glucose by Outcome
sns.kdeplot(df[df['Outcome']==0]['Glucose'], ax=axes[0], label='No Diabetes', color='green', fill=True, alpha=0.3)
sns.kdeplot(df[df['Outcome']==1]['Glucose'], ax=axes[0], label='Diabetes',    color='red',   fill=True, alpha=0.3)
axes[0].set_title('KDE of Glucose by Outcome')
axes[0].legend()

# Plot 2: Boxplot of BMI by Outcome
sns.boxplot(data=df, x='Outcome', y='BMI', ax=axes[1], palette={0:'lightblue', 1:'lightsalmon'})
axes[1].set_title('BMI by Outcome (Boxplot)')

# Plot 3: Scatter Age vs Glucose colored by Outcome
colors = df['Outcome'].map({0: 'green', 1: 'red'})
axes[2].scatter(df['Age'], df['Glucose'], c=colors, alpha=0.4, edgecolors='k', linewidths=0.2)
axes[2].set_title('Age vs Glucose (by Outcome)')
axes[2].set_xlabel('Age')
axes[2].set_ylabel('Glucose')

plt.suptitle('Bonus Challenge — Three Views of Diabetes Data', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
