```python
# prices = [150, 80, 320, 45, 500, 260, 99]
# a) Select all prices above 200
# b) Select all prices between 50 and 300 (use & operator)
# c) Count how many items cost less than 100

import numpy as np

prices = np.array([150, 80, 320, 45, 500, 260, 99])

above_200 = prices[prices > 200]
print("Prices above 200:", above_200)

between_50_300 = prices[(prices >= 50) & (prices <= 300)]
print("Prices between 50 and 300:", between_50_300)

less_than_100 = np.sum(prices < 100)
print("Count of items costing less than 100:", less_than_100)
```

```python
# sales = [1200, 450, 890, 2300, 670, 1800, 300, 950]
# a) Select sales above the average (use np.mean)
# b) Replace all sales below 500 with 0 (use boolean assignment)
# c) Print the modified array

import numpy as np

sales = np.array([1200, 450, 890, 2300, 670, 1800, 300, 950])

avg_sales = np.mean(sales)

above_avg = sales[sales > avg_sales]
print("Sales above average:", above_avg)
print("Average sales:", avg_sales)

sales[sales < 500] = 0
print("Modified array (sales < 500 replaced with 0):", sales)
```

```python
# exam_marks = [45.9, 78.3, 92.7, 33.1, 60.5]
# a) Cast to int (observe what happens to decimals)
# b) Use np.round() first, THEN cast to int
# c) Print both and explain the difference

import numpy as np

exam_marks = np.array([45.9, 78.3, 92.7, 33.1, 60.5])

# Direct cast to int (truncates decimals)
marks_direct_int = exam_marks.astype(int)
print("Direct cast to int (truncated):", marks_direct_int)

# Round first, then cast to int
marks_rounded_int = np.round(exam_marks).astype(int)
print("Round first then cast to int:", marks_rounded_int)

print("\nExplanation: Direct casting truncates decimals (floor towards zero),")
print("while rounding first gives the nearest integer value.")
```

```python
# initial_investment = [1000, 2000, 5000, 800]
# annual_return_rate = [0.08, 0.12, 0.05, 0.15]
# Calculate value after 1 year: value = investment * (1 + rate)
# Calculate profit = value - initial_investment

import numpy as np

initial_investment = np.array([1000, 2000, 5000, 800])
annual_return_rate = np.array([0.08, 0.12, 0.05, 0.15])

value_after_1_year = initial_investment * (1 + annual_return_rate)
profit = value_after_1_year - initial_investment

print("Initial investment:", initial_investment)
print("Annual return rate:", annual_return_rate)
print("Value after 1 year:", value_after_1_year)
print("Profit:", profit)
```
