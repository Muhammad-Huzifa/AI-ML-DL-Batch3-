```python
# Introduction to NumPy - All Exercises Solved


# Hour 1 — Introduction to NumPy Exercises

# Exercise 1: Import numpy and print its version
import numpy as np
print(np.__version__)

# Exercise 2: Create a Python list of 5 student marks and convert it into a NumPy array
marks_list = [75, 82, 90, 68, 85]
marks_array = np.array(marks_list)
print("Python list:", marks_list)
print("Type of list:", type(marks_list))
print("NumPy array:", marks_array)
print("Type of array:", type(marks_array))

# Exercise 3: Compare speed of multiplying 500,000 elements using Python list vs NumPy array
import time
size = 500000
py_list = list(range(size))
np_array = np.arange(size)
start = time.time()
result_list = [x * 2 for x in py_list]
list_time = time.time() - start
start = time.time()
result_np = np_array * 2
np_time = time.time() - start
print(f"Python list time: {list_time:.4f} seconds")
print(f"NumPy array time: {np_time:.4f} seconds")
print(f"NumPy is {list_time/np_time:.1f}x faster!")

# Exercise 4: Create a NumPy array of temperatures
temperatures = np.array([36.5, 37.1, 36.8, 37.4, 38.0])
print("Temperatures array:", temperatures)
print("Data type:", temperatures.dtype)

# Part A: From Python Lists Exercises
# Exercise 1: 1D NumPy array of 6 product prices
prices = np.array([299.99, 149.50, 89.99, 499.00, 24.99, 12.50])
print("Product prices:", prices)

# Exercise 2: 2D NumPy array representing a 3x3 multiplication table
mult_table = np.array([[1, 2, 3], [2, 4, 6], [3, 6, 9]])
print("Multiplication table:\n", mult_table)
numbers = np.arange(1, 4)
mult_table_broadcast = numbers.reshape(3, 1) * numbers.reshape(1, 3)
print("\nUsing broadcasting:\n", mult_table_broadcast)

# Exercise 3: Mixed list array
mixed_list = [1, 2, 3.5, 4]
mixed_array = np.array(mixed_list)
print("Mixed array:", mixed_array)
print("Data type:", mixed_array.dtype)

# Exercise 4: 2D array for 4 students (Math, Science)
student_marks = np.array([[85, 90], [78, 82], [92, 88], [70, 75]])
print("Student marks (Math, Science):\n", student_marks)
print("Shape:", student_marks.shape)

# Part B: Built-in Methods Exercises
# Exercise 1: 4x4 zeros array
empty_scores = np.zeros((4, 4))
print("Empty score table (4x4 zeros):\n", empty_scores)

# Exercise 2: Roll numbers 1 to 30
roll_numbers = np.arange(1, 31)
print("Roll numbers 1 to 30:", roll_numbers)

# Exercise 3: Linspace 20 to 100
temperatures = np.linspace(20, 100, 8)
print("8 equally spaced temperatures from 20 to 100:", temperatures)

# Exercise 4: Full array 3x3 with 99
full_array = np.full((3, 3), 99)
print("3x3 array filled with 99:\n", full_array)

# Part C: Random Arrays Exercises
# Exercise 1: Random exam scores 50 to 100
np.random.seed(42)
exam_scores = np.random.randint(50, 101, size=10)
print("10 random exam scores:", exam_scores)

# Exercise 2: 4x4 random matrix 0 to 1
np.random.seed(42)
random_matrix = np.random.rand(4, 4)
print("4x4 random matrix (0 to 1):\n", random_matrix)

# Exercise 3: Seed 10 reproducibility
np.random.seed(10)
first_run = np.random.rand(5)
print("First run with seed(10):", first_run)
np.random.seed(10)
second_run = np.random.rand(5)
print("Second run with seed(10):", second_run)
print("Are they identical?", np.array_equal(first_run, second_run))

# Exercise 4: Simulate 7 daily body temperatures
np.random.seed(42)
body_temps = 36.5 + 0.3 * np.random.randn(7)
print("7 simulated daily body temperatures:", body_temps)
print("Rounded to 1 decimal:", np.round(body_temps, 1))

# Array Attributes and Methods Exercises
np.random.seed(42)
marks = np.random.randint(40, 101, size=(5, 3))
print("Marks array (5 students × 3 subjects):\n", marks)
print("\nShape:", marks.shape)
print("Size:", marks.size)
print("Dimensions (ndim):", marks.ndim)
print("Data type:", marks.dtype)

# Exercise 2: Reshape
arr_1d = np.arange(1, 13)
print("Original 1D array:", arr_1d)
arr_3x4 = arr_1d.reshape(3, 4)
print("\nReshaped to (3, 4):\n", arr_3x4)
arr_2x6 = arr_1d.reshape(2, 6)
print("\nReshaped to (2, 6):\n", arr_2x6)

# Exercise 3: Max and argmax
np.random.seed(42)
scores = np.random.randint(50, 101, size=8)
print("Exam scores:", scores)
highest_score = scores.max()
student_index = scores.argmax()
print(f"Highest score: {highest_score}")
print(f"Student at index {student_index} got the highest score")

# Exercise 4: 3D array
arr_3d = np.arange(24).reshape(2, 3, 4)
print("3D array:\n", arr_3d)
print("\nNumber of dimensions (ndim):", arr_3d.ndim)
print("Shape:", arr_3d.shape)
print("Size (total elements):", arr_3d.size)

# Copying Exercises
# Exercise 1
scores = np.array([85, 90, 78, 92, 88])
scores_copy = scores.copy()
scores_copy[2] = 100
print("Original scores:", scores)
print("Copy after modification:", scores_copy)
print("Original unchanged? Original[2] =", scores[2])

# Exercise 2
arr = np.array([10, 20, 30, 40, 50])
arr_view = arr.view()
arr_view[0] = 999
print("Original after view change:", arr)
print("View:", arr_view)

# Exercise 3
original = np.array([5, 10, 15, 20, 25])
copy1 = original.copy()
copy2 = original.copy()
copy1 = copy1 + 10
copy2 = copy2 - 10
print("Original:", original)
print("Copy1 (+10):", copy1)
print("Copy2 (-10):", copy2)

# Exercise 4
arr = np.array([1, 2, 3, 4, 5])
arr_copy = arr.copy()
arr_view = arr.view()
print("Does copy share memory with original?", np.shares_memory(arr, arr_copy))
print("Does view share memory with original?", np.shares_memory(arr, arr_view))

# Append and Insert Exercises
# Exercise 1
prices = np.array([199, 299, 399, 99, 149])
new_prices = np.append(prices, 249)
print("Original prices:", prices)
print("After appending 249:", new_prices)

# Exercise 2
roll_numbers = np.array([1, 2, 4, 5])
roll_numbers_updated = np.insert(roll_numbers, 2, 3)
print("Original roll numbers:", roll_numbers)
print("After inserting 3 at index 2:", roll_numbers_updated)

# Exercise 3
scores = np.array([85, 90, 78, 95])
new_scores = np.append(scores, [72, 68, 91])
print("Original scores:", scores)
print("After appending multiple scores:", new_scores)

# Exercise 4
arr = np.array([10, 20, 30, 40, 50])
arr_with_begin = np.insert(arr, 0, 0)
arr_with_both = np.insert(arr_with_begin, len(arr_with_begin), 0)
print("Original:", arr)
print("After inserting 0 at beginning:", arr_with_begin)
print("After inserting 0 at end also:", arr_with_both)

# Sorting Arrays Exercises
np.random.seed(42)
scores = np.random.randint(50, 101, size=8)
sorted_scores = np.sort(scores)
print("Unsorted scores:", scores)
print("Sorted ascending:", sorted_scores)

prices = np.array([299, 99, 599, 199, 399, 49, 799])
sorted_desc = np.sort(prices)[::-1]
print("Original prices:", prices)
print("Sorted descending:", sorted_desc)

np.random.seed(42)
marks_table = np.random.randint(40, 101, size=(3, 4))
sorted_by_row = np.sort(marks_table, axis=1)
print("Original marks table (3 students × 4 subjects):\n", marks_table)
print("\nSorted each student's marks (along axis=1):\n", sorted_by_row)

arr = np.array([5, 2, 8, 1, 9, 3])
sorted_arr = np.sort(arr)
print("Before sorting:", arr)
print("After sorting:", sorted_arr)
print("Original array unchanged:", arr)

# Removing/Deleting Elements Exercises
scores = np.array([85, 90, 78, 92, 88, 95])
scores_updated = np.delete(scores, 4)
print("All scores:", scores)
print("After removing index 4 (student absent):", scores_updated)

temps = np.array([36.5, 37.1, 36.8, 37.4, 38.0, 36.9, 37.2, 36.7])
temps_trimmed = np.delete(temps, [0, -1])
print("Original temperatures:", temps)
print("After removing first and last:", temps_trimmed)

marks_table = np.array([[85, 90, 78], [70, 88, 92], [95, 82, 89]])
marks_updated = np.delete(marks_table, 1, axis=0)
print("Original marks table:\n", marks_table)
print("\nAfter deleting row 1 (student withdrew):\n", marks_updated)

arr = np.array([10, 20, 30, 40, 50, 60, 70])
arr_deleted = np.delete(arr, [1, 3, 5])
print("Original array:", arr)
print("After deleting indices 1, 3, 5:", arr_deleted)

# Combining/Concatenating Arrays Exercises
section_a = np.array([80, 85, 90])
section_b = np.array([70, 75, 88])
all_scores = np.concatenate([section_a, section_b])
print("Section A scores:", section_a)
print("Section B scores:", section_b)
print("Combined scores:", all_scores)

student1 = np.array([[85, 90, 78]])
student2 = np.array([[70, 88, 92]])
marks_table = np.vstack([student1, student2])
print("Student 1 marks:", student1)
print("Student 2 marks:", student2)
print("\nVertically stacked (2x3 table):\n", marks_table)

math_marks = np.array([[85, 90], [78, 88]])
science_marks = np.array([[92, 86], [80, 94]])
combined = np.hstack([math_marks, science_marks])
print("Math marks (2x2):\n", math_marks)
print("\nScience marks (2x2):\n", science_marks)
print("\nHorizontally stacked (2x4):\n", combined)

arr1 = np.array([1, 2])
arr2 = np.array([3, 4])
arr3 = np.array([5, 6])
combined = np.concatenate([arr1, arr2, arr3])
print("Array 1:", arr1)
print("Array 2:", arr2)
print("Array 3:", arr3)
print("Combined:", combined)
print("Final shape:", combined.shape)
print("Final size:", combined.size)

# Splitting Arrays Exercises
roll_numbers = np.arange(1, 13)
groups = np.split(roll_numbers, 4)
print("Roll numbers:", roll_numbers)
for i, group in enumerate(groups):
    print(f"  Group {i+1}: {group}")

scores = np.arange(1, 11)
groups = np.array_split(scores, 3)
print("Scores:", scores)
for i, group in enumerate(groups):
    print(f"  Group {i+1}: {group} (size: {len(group)})")

marks_table = np.arange(1, 19).reshape(6, 3)
top_half, bottom_half = np.vsplit(marks_table, 2)
print("Marks table (6x3):\n", marks_table)
print("\nTop half (first 3 rows):\n", top_half)
print("\nBottom half (last 3 rows):\n", bottom_half)

arr = np.arange(1, 19).reshape(3, 6)
parts = np.hsplit(arr, 3)
print("Original array (3x6):\n", arr)
for i, part in enumerate(parts):
    print(f"\nPart {i+1} (3x2):\n", part)

```
