## 1. Data Loading & Saving
**`np.save()` stores a NumPy array to disk; `np.load()` reads it back.**

```python
import numpy as np

arr = np.arange(12).reshape(3, 4)
print(arr)
np.save('my_array.npy', arr)     
     # File extension .npy is added automatically if missing

arr2=np.arange(16).reshape(4,4)
np.save('numpy.npy',arr2)

loaded=np.load('numpy.npy')
print(loaded)
```

**Accessing elements, rows, and columns from arrays by position or condition.**

```python
arr2d = np.array([[10, 20, 30],
                  [40, 50, 60],
                  [70, 80, 90]])

print('Single element:', arr2d[1, 1])
print('Full row 0:    ', arr2d[0])
print('Full col 2:    ', arr2d[:, 2])
print('Sub-array:\n',    arr2d[0:2, 0:2])
```

## 3. Broadcasting
**NumPy automatically expands smaller arrays to match bigger ones during operations.**

```python
arr = np.array([1, 2, 3, 4])
print('Array + 10:  ', arr + 10)
print('Array * 3:   ', arr * 3)
print('Array ** 2:  ', arr ** 2)
```

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
row    = np.array([10, 20, 30])

print('Matrix + row:\n', matrix + row)
print('Matrix shapes:', matrix.shape, '+', row.shape)
```

#### Scalar

```python
M = np.array([
[1,2],
[3,4],
[5,6]
])

M + 5
```

##### Row Vector

```python
N = np.array([
[10,20,30],
[40,50,60]
])
v = np.array([1,2,3])

N + v
```

## 4. Type Casting
**`.astype()` converts array elements from one data type to another.**

```python
float_arr = np.array([1.9, 2.7, 3.3, 4.8])
int_arr   = float_arr.astype(int)
print('float:', float_arr)
print('int:  ', int_arr)
```

```python
int_arr2   = np.array([1, 0, 1, 0, 1])
bool_arr   = int_arr2.astype(bool)
float_arr2 = int_arr2.astype(float)
print('int:  ', int_arr2)
print('bool: ', bool_arr)
print('float:', float_arr2)
```

**Practice 1 — Type Casting**

```python
sensor_readings = np.array([98.6, 101.2, 99.8, 103.5])
print('dtype before:', sensor_readings.dtype)
as_int = sensor_readings.astype(np.int32)
print('as int32:', as_int)
```

```python
flags = np.array([1, 0, 0, 1, 1, 0])
as_bool = flags.astype(bool)
print('flags as bool:', as_bool)
print('sum of True:', np.sum(as_bool))
```

## 5. Arithmetic Operations
**Element-wise: add, subtract, multiply, divide, exponentiation.**

```python
a = np.array([10, 20, 30, 40])
b = np.array([1,  2,  3,  4 ])

print('Add:        ', a + b)
print('Subtract:   ', a - b)
print('Multiply:   ', a * b)
print('Divide:     ', a / b)
print('Exponent:   ', b ** 3)
```

```python
income   = np.array([50000, 75000, 30000, 90000])
expenses = np.array([20000, 40000, 15000, 60000])

savings      = income - expenses
savings_pct  = (savings / income) * 100
print('Savings:    ', savings)
print('Savings %:  ', savings_pct)
```

## 6. Universal Array Functions (ufuncs)
**Built-in math functions that operate on entire arrays: sqrt, exp, sin, log, max, etc.**

```python
arr = np.array([1, 4, 9, 16, 25])
print('sqrt:', np.sqrt(arr))
print('max: ', np.max(arr))
print('min: ', np.min(arr))
print('sum: ', np.sum(arr))
print('mean:', np.mean(arr))
```

```python
angles = np.array([0, np.pi/6, np.pi/4, np.pi/2])
print('sin:', np.round(np.sin(angles), 2))
print('cos:', np.round(np.cos(angles), 2))

x = np.array([0, 1, 2, 3])
print('exp:', np.round(np.exp(x), 3))
print('log:', np.round(np.log(np.exp(x)), 3))
```

**Practice 1 — Universal Functions**

```python
areas = np.array([4.0, 9.0, 16.0, 25.0, 36.0])
sides = np.sqrt(areas)
print('Area  → Side')
for a, s in zip(areas, sides):
    print(f'  {a:.1f}  → {s:.1f}')
```

```python
temperatures = np.array([-5, 0, 15, 22, 35, 42, -10, 8])
print('Max temp:   ', np.max(temperatures))
print('Min temp:   ', np.min(temperatures))
print('Mean temp:  ', np.mean(temperatures))
print('Abs values: ', np.abs(temperatures))
```
