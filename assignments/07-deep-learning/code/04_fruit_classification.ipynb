## 🏗️ Task 3 — Build the Transfer Learning Model (25 Marks)

### Background
**Transfer Learning** reuses a model pre-trained on a large dataset (ImageNet — 1.2M images, 1000 classes) as a feature extractor for our new task. ResNet50 has already learned powerful low-level features (edges, textures) and high-level features (shapes, objects) that are useful for fruit classification.

### Architecture Overview
```
Input (224×224×3)
    ↓
Data Augmentation
    ↓
Pixel Rescaling (÷255)
    ↓
ResNet50 Base (frozen weights — pretrained on ImageNet)
    ↓
GlobalAveragePooling2D
    ↓
BatchNormalization
    ↓
Dense(256, relu) → Dropout
    ↓
Dense(128, relu) → Dropout
    ↓
Dense(8, softmax)  ← 8 fruit classes
```

- `GlobalAveragePooling2D` converts `(batch, h, w, channels)` → `(batch, channels)` by averaging spatially.
- `Dropout(rate)` randomly sets `rate` fraction of activations to zero during training.

## 🔓 Task 5 — Phase 2: Fine-Tuning (15 Marks)

### Background
In **Phase 2 (Fine-Tuning)**, we **unfreeze** the top layers of ResNet50 and train them with a **very small learning rate**. This allows the deep features to be slightly adjusted for our specific fruit classification task.

**Why only top layers?**
- Early layers learn simple, universal features (edges, colors) — these are already perfect.
- Later layers learn task-specific features — these benefit from fine-tuning.
- Unfreezing all layers risks **catastrophic forgetting**.

### 💡 Hints
- Set `base_model.trainable = True` to unfreeze all layers first.
- Then loop: `for layer in base_model.layers[:-N]: layer.trainable = False` to keep early layers frozen.
- A good value for `N` (layers to unfreeze from the top) is **20–40**.
- **Important**: After changing trainable status, you **must recompile** the model.
- Use a **10x smaller learning rate** than Phase 1 (e.g., `0.0001`) to avoid destroying pre-trained weights.
