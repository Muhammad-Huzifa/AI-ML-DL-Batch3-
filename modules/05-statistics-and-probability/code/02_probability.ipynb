## Hour 1–2 : Measures of Dispersion

### Variance
Variance = average of squared differences from the mean.

````python
data = [10, 12, 14, 16, 18]

mean = sum(data) / len(data)
variance = sum((x - mean) ** 2 for x in data) / len(data)

print('Data    :', data)
print('Mean    :', mean)
print('Variance:', variance)
````

### Standard Deviation
Standard Deviation = square root of variance. Same unit as the original data.

````python
data = [10, 12, 14, 16, 18]

mean = sum(data) / len(data)
variance = sum((x - mean) ** 2 for x in data) / len(data)
std_dev = variance ** 0.5

print('Data               :', data)
print('Mean               :', mean)
print('Variance           :', variance)
print('Standard Deviation :', round(std_dev, 2))
````

### Coefficient of Variation (CV)
CV = (Standard Deviation ÷ Mean) × 100. Expresses spread as a percentage — unit-free.

````python
salaries = [30, 35, 40, 45, 50]

mean = sum(salaries) / len(salaries)
variance = sum((x - mean) ** 2 for x in salaries) / len(salaries)
std_dev = variance ** 0.5
cv = (std_dev / mean) * 100

print('Salaries (000s PKR):', salaries)
print('Mean               :', mean)
print('Std Dev            :', round(std_dev, 2))
print('CV                 :', round(cv, 2), '%')
````

### Skewness
Skewness describes whether data leans left (negative), is symmetric (zero), or leans right (positive).

````python
import statistics

symmetric     = [10, 20, 30, 40, 50]
right_skewed  = [5, 6, 7, 8, 100]
left_skewed   = [1, 90, 92, 95, 98]

for label, dataset in [('Symmetric', symmetric), ('Right Skewed', right_skewed), ('Left Skewed', left_skewed)]:
    mean   = sum(dataset) / len(dataset)
    median = statistics.median(dataset)
    print(f'{label:15} → Mean: {mean:.1f}  Median: {median}  (Mean vs Median tells the direction)')
````

### Kurtosis
Kurtosis measures the sharpness of the peak and the weight of the tails. Normal distribution kurtosis = 3.

```Kurtosis = Σ[(x - μ)⁴] / (n × σ⁴)```

## Hour 3–4 : Measures of Position

### Z-Score
Z = (value − mean) ÷ standard deviation. Tells how many SDs a value is from the mean.

````python
scores = [60, 70, 80, 85, 90]
student_score = 85

mean    = sum(scores) / len(scores)
variance = sum((x - mean) ** 2 for x in scores) / len(scores)
std_dev  = variance ** 0.5
z_score  = (student_score - mean) / std_dev

print('Scores        :', scores)
print('Mean          :', mean)
print('Std Dev       :', round(std_dev, 2))
print('Student Score :', student_score)
print('Z-Score       :', round(z_score, 2))
````

### Percentile
Percentile rank = (number of values below ÷ total values) × 100.

````python
scores = [45, 52, 58, 63, 67, 72, 78, 85, 90, 96]
target = 78

below  = sum(1 for s in scores if s < target)
percentile = (below / len(scores)) * 100

print('Scores    :', scores)
print('Target    :', target)
print('Below     :', below)
print('Percentile:', percentile)
````

### Quartile (Q1, Q2, Q3) and IQR
Q1=25th percentile, Q2=median (50th), Q3=75th percentile. IQR = Q3 − Q1.

````python
data = sorted([4, 8, 15, 16, 23, 42, 55, 60, 72, 88])
n = len(data)

q2 = (data[n // 2 - 1] + data[n // 2]) / 2

lower_half = data[:n // 2]
upper_half = data[n // 2:]

q1 = lower_half[len(lower_half) // 2]
q3 = upper_half[len(upper_half) // 2]
iqr = q3 - q1

print('Data (sorted):', data)
print('Q1 (25th)    :', q1)
print('Q2 (Median)  :', q2)
print('Q3 (75th)    :', q3)
print('IQR          :', iqr)
````
