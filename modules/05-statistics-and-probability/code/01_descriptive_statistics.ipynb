## Hour 1–2 : Data & Its Types

### Structured vs Unstructured Data
Structured data is organised in rows/columns. Unstructured has no fixed format.

```python
structured = {
    'Name':  ['Ali', 'Sara', 'Ahmed'],
    'Age':   [22, 25, 28],
    'Score': [85, 90, 78]
}

unstructured = [
    'Hey! Can we meet tomorrow?',
    'Image: photo_holiday.jpg',
    'Voice message received at 9pm'
]

print('Structured Data:')
for key, values in structured.items():
    print(f'  {key}: {values}')

print('\nUnstructured Data:')
for item in unstructured:
    print(f'  {item}')
```

### Quantitative Data — Continuous vs Discrete
Continuous: measured values (decimals allowed). Discrete: counted values (whole numbers only).

## Hour 3–4 : Measures of Central Tendency

### Mean
Mean = Sum of all values ÷ Total count.

```python
marks = [60, 70, 80, 90, 100]

mean = sum(marks) / len(marks)

print('Marks:', marks)
print('Mean :', mean)
```

### Median
Median = middle value after sorting. If even count → average of two middle values.

```python
data = [8, 3, 5, 1, 7]

data_sorted = sorted(data)
n = len(data_sorted)

if n % 2 != 0:
    median = data_sorted[n // 2]
else:
    median = (data_sorted[n // 2 - 1] + data_sorted[n // 2]) / 2

print('Original :', data)
print('Sorted   :', data_sorted)
print('Median   :', median)
```

### Mode
Mode = the value that appears most often.

```python
from collections import Counter

scores = [4, 7, 2, 7, 3, 7, 5, 2]

counts = Counter(scores)
mode   = counts.most_common(1)[0][0]

print('Scores:', scores)
print('Mode  :', mode)
```

### Mean, Median & Mode Together
Calculating all three measures on the same dataset.

```python
from collections import Counter

salaries = [30000, 32000, 35000, 33000, 32000, 200000]

mean   = sum(salaries) / len(salaries)

s = sorted(salaries)
n = len(s)
median = s[n // 2] if n % 2 != 0 else (s[n // 2 - 1] + s[n // 2]) / 2

mode   = Counter(salaries).most_common(1)[0][0]

print('Salaries:', salaries)
print('Mean    :', mean)
print('Median  :', median)
print('Mode    :', mode)
print('\nNotice: Mean is pulled up by 200000 — Median is more realistic here!')
```
