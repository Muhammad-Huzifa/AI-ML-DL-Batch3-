## 2. Discretization and Binning

```python
# Cut Age into 4 equal-width bins
df['Age_Bin'] = pd.cut(df['Age'], bins=4)
df[['Age', 'Age_Bin']].head(10)
```

```python
# Cut Age into 4 bins with custom labels
df['Age_Group'] = pd.cut(df['Age'], bins=4, labels=['Young', 'Adult', 'MiddleAge', 'Senior'])
df[['Age', 'Age_Group']].head(10)
```

```python
# Count how many rows fall in each Age_Group
df['Age_Group'].value_counts()
```

```python
# Cut BMI into 5 equal-width bins
df['BMI_Bin'] = pd.cut(df['BMI'], bins=5)
df['BMI_Bin'].value_counts().sort_index()
```

```python
# qcut divides into equal-frequency quantile bins
df['Glucose_Quantile'] = pd.qcut(df['Glucose'], q=4, labels=['Low', 'MedLow', 'MedHigh', 'High'])
df['Glucose_Quantile'].value_counts()
```

```python
# Use custom bin edges for Blood Pressure
df['BP_Category'] = pd.cut(df['BloodPressure'], bins=[0, 60, 80, 90, 200], labels=['Low', 'Normal', 'High', 'VeryHigh'])
df['BP_Category'].value_counts()
```

## 5. Matplotlib Basics (All Core Functions)
> Every fundamental matplotlib function students need to know

```python
# ── Figure and Axes: the two main objects in matplotlib ──
fig, ax = plt.subplots()          # create one figure with one plot area
ax.plot([1, 2, 3], [4, 5, 6])    # draw a line on the axes
ax.set_title('My First Plot')     # add a title
plt.show()
```

```python
# ── plt.figure: control figure size and DPI ──
plt.figure(figsize=(8, 4), dpi=100)    # width=8 inches, height=4 inches
plt.plot([1, 2, 3], [10, 20, 15])
plt.title('Figure Size Demo')
plt.show()
```

```python
# ── plt.subplots: create multiple plots side by side ──
fig, axes = plt.subplots(1, 2, figsize=(12, 4))   # 1 row, 2 columns
axes[0].plot(df['Age'], df['Glucose'], 'b.')       # left plot
axes[0].set_title('Age vs Glucose')
axes[1].plot(df['BMI'], df['Glucose'], 'r.')       # right plot
axes[1].set_title('BMI vs Glucose')
plt.tight_layout()    # prevent overlap between subplots
plt.show()
```

```python
# ── plt.plot: basic line plot ──
plt.figure(figsize=(8, 4))
plt.plot(df['Glucose'].values[:50])           # just pass values, index becomes x
plt.title('Glucose Values (first 50 rows)')
plt.xlabel('Row Index')                        # label x-axis
plt.ylabel('Glucose')                          # label y-axis
plt.show()
```

```python
# ── Line style, color, marker options ──
plt.figure(figsize=(8, 4))
plt.plot(df['Glucose'].values[:30], color='red', linestyle='--', marker='o', linewidth=2, markersize=5)
plt.title('Line Style Options')
plt.grid(True)       # show grid lines
plt.show()
```

```python
# ── plt.scatter: scatter plot (two numeric columns) ──
plt.figure(figsize=(7, 5))
plt.scatter(df['Age'], df['Glucose'], alpha=0.5, color='blue', edgecolors='black', s=30)
# alpha = transparency, s = dot size
plt.title('Age vs Glucose')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.show()
```

```python
# ── Scatter with color by class ──
plt.figure(figsize=(7, 5))
colors = df['Outcome'].map({0: 'green', 1: 'red'})    # map 0→green, 1→red
plt.scatter(df['Age'], df['Glucose'], c=colors, alpha=0.5, s=30)
plt.title('Age vs Glucose (colored by Outcome)')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.show()
```

```python
# ── plt.bar: vertical bar chart ──
counts = df['Outcome'].value_counts()    # count 0s and 1s
plt.figure(figsize=(6, 4))
plt.bar(counts.index, counts.values, color=['steelblue', 'salmon'], edgecolor='black', width=0.5)
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Outcome Count')
plt.ylabel('Count')
plt.show()
```

```python
# ── plt.barh: horizontal bar chart ──
mean_vals = df.drop('Outcome', axis=1).mean()   # mean of each feature
plt.figure(figsize=(8, 5))
plt.barh(mean_vals.index, mean_vals.values, color='teal')
plt.title('Mean Value of Each Feature')
plt.xlabel('Mean')
plt.show()
```

```python
# ── plt.hist: histogram (distribution of one column) ──
plt.figure(figsize=(7, 4))
plt.hist(df['Glucose'], bins=20, color='steelblue', edgecolor='black')
plt.title('Glucose Distribution')
plt.xlabel('Glucose')
plt.ylabel('Frequency')
plt.show()
```

```python
# ── Two histograms overlapping ──
plt.figure(figsize=(8, 4))
plt.hist(df[df['Outcome']==0]['Glucose'], bins=20, alpha=0.6, color='green', label='No Diabetes')
plt.hist(df[df['Outcome']==1]['Glucose'], bins=20, alpha=0.6, color='red',   label='Diabetes')
plt.legend()    # show legend box
plt.title('Glucose by Outcome')
plt.xlabel('Glucose')
plt.show()
```

```python
# ── plt.boxplot: box-and-whisker plot ──
plt.figure(figsize=(7, 5))
plt.boxplot([df[df['Outcome']==0]['BMI'], df[df['Outcome']==1]['BMI']],
            labels=['No Diabetes', 'Diabetes'])
plt.title('BMI Distribution by Outcome')
plt.ylabel('BMI')
plt.show()
```

```python
# ── plt.pie: pie chart ──
counts = df['Outcome'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(counts.values, labels=['No Diabetes', 'Diabetes'],
        autopct='%1.1f%%', colors=['lightblue', 'salmon'],
        startangle=90, explode=[0, 0.05])
# autopct shows percentage, explode separates a slice
plt.title('Outcome Distribution')
plt.show()
```

```python
# ── plt.legend, plt.grid, plt.xlim, plt.ylim ──
plt.figure(figsize=(8, 4))
plt.plot(df['Glucose'].values[:50], label='Glucose', color='blue')
plt.plot(df['BloodPressure'].values[:50], label='Blood Pressure', color='red')
plt.legend(loc='upper right')    # place legend
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(0, 50)                  # set x-axis range
plt.ylim(0, 200)                 # set y-axis range
plt.title('Glucose vs BloodPressure')
plt.show()
```

```python
# ── plt.axhline / plt.axvline: reference lines ──
plt.figure(figsize=(8, 4))
plt.hist(df['Glucose'], bins=20, color='steelblue', edgecolor='black')
plt.axvline(df['Glucose'].mean(), color='red',    linestyle='--', label='Mean')
plt.axvline(df['Glucose'].median(), color='green', linestyle=':',  label='Median')
plt.legend()
plt.title('Glucose with Mean and Median Lines')
plt.show()
```

```python
# ── plt.text / plt.annotate: add text on plot ──
plt.figure(figsize=(7, 5))
plt.scatter(df['Age'], df['Glucose'], alpha=0.4, s=20)
plt.text(70, 190, 'High Risk Zone', fontsize=12, color='red')   # add raw text
plt.annotate('Max Glucose', xy=(df['Age'][df['Glucose'].idxmax()], df['Glucose'].max()),
             xytext=(60, 180), arrowprops=dict(arrowstyle='->'), fontsize=10)
plt.title('Annotated Scatter')
plt.show()
```

```python
# ── plt.xticks / plt.yticks: control tick labels ──
plt.figure(figsize=(7, 4))
plt.hist(df['Age'], bins=10, color='purple', edgecolor='black')
plt.xticks(range(20, 85, 5), rotation=45)    # custom ticks every 5 years
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# ── plt.savefig: save the figure to a file ──
plt.figure(figsize=(7, 4))
plt.hist(df['BMI'], bins=20, color='orange', edgecolor='black')
plt.title('BMI Distribution')
plt.savefig('bmi_histogram.png', dpi=150, bbox_inches='tight')
# bbox_inches='tight' prevents cutting off labels
plt.show()
print('Figure saved as bmi_histogram.png')
```

```python
# ── plt.subplots 2x2 grid: all 4 features at once ──
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0,0].hist(df['Glucose'], bins=20, color='steelblue', edgecolor='black')
axes[0,0].set_title('Glucose')
axes[0,1].hist(df['BMI'], bins=20, color='salmon', edgecolor='black')
axes[0,1].set_title('BMI')
axes[1,0].hist(df['Age'], bins=20, color='green', edgecolor='black')
axes[1,0].set_title('Age')
axes[1,1].hist(df['Insulin'], bins=20, color='purple', edgecolor='black')
axes[1,1].set_title('Insulin')
plt.suptitle('Feature Distributions', fontsize=14)   # main title for all
plt.tight_layout()
plt.show()
```

## 8. Distribution Plots

```python
# distplot: histogram + KDE curve together (classic seaborn)
plt.figure(figsize=(7, 4))
sns.histplot(df['Glucose'], kde=True, color='steelblue', bins=20)
# histplot replaces deprecated distplot in modern seaborn
plt.title('Glucose Distribution (histplot + KDE)')
plt.show()
```

```python
# kdeplot: smooth density curve only
plt.figure(figsize=(7, 4))
sns.kdeplot(df['Glucose'], fill=True, color='purple', alpha=0.5)
plt.title('Glucose KDE Plot')
plt.show()
```

```python
# kdeplot: compare two groups on same axis
plt.figure(figsize=(8, 4))
sns.kdeplot(df[df['Outcome']==0]['Glucose'], label='No Diabetes', fill=True, alpha=0.4)
sns.kdeplot(df[df['Outcome']==1]['Glucose'], label='Diabetes',    fill=True, alpha=0.4)
plt.legend()
plt.title('Glucose KDE by Outcome')
plt.show()
```

```python
# rugplot: tiny tick marks at each data point along the axis
plt.figure(figsize=(8, 3))
sns.rugplot(df['Glucose'], height=0.5, color='steelblue')
plt.title('Glucose Rugplot')
plt.show()
```

```python
# jointplot: scatter + distribution on both axes
sns.jointplot(data=df, x='Age', y='Glucose', kind='scatter', height=6)
plt.suptitle('Age vs Glucose Jointplot', y=1.02)
plt.show()
```

```python
# jointplot kind='hex': hexbin density version
sns.jointplot(data=df, x='BMI', y='Glucose', kind='hex', height=6)
plt.suptitle('BMI vs Glucose (Hexbin)', y=1.02)
plt.show()
```

```python
# jointplot kind='kde': density contours
sns.jointplot(data=df, x='Age', y='BMI', kind='kde', height=6, fill=True)
plt.suptitle('Age vs BMI (KDE)', y=1.02)
plt.show()
```

```python
# pairplot: scatter matrix of all numeric columns
sns.pairplot(df, hue='Outcome', palette={0: 'green', 1: 'red'}, plot_kws={'alpha': 0.4})
plt.suptitle('Pairplot of All Features', y=1.02)
plt.show()
```

```python
# pairplot with only selected columns
sns.pairplot(df[['Glucose', 'BMI', 'Age', 'Outcome']], hue='Outcome', diag_kind='kde')
plt.suptitle('Pairplot: Glucose, BMI, Age', y=1.02)
plt.show()
```

## 9. Categorical Data Plots

```python
# boxplot: median, quartiles, outliers per category
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Outcome', y='Glucose', palette='Set2')
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Glucose by Outcome (Boxplot)')
plt.show()
```

```python
# boxplot: multiple features side by side
plt.figure(figsize=(12, 5))
sns.boxplot(data=df.drop('Outcome', axis=1), palette='pastel')
plt.title('Boxplot of All Features')
plt.xticks(rotation=30)
plt.show()
```

```python
# violinplot: like boxplot but also shows density shape
plt.figure(figsize=(7, 5))
sns.violinplot(data=df, x='Outcome', y='BMI', palette='muted')
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('BMI by Outcome (Violinplot)')
plt.show()
```

```python
# violinplot with split: compare both outcomes in one violin
plt.figure(figsize=(12, 5))
sns.violinplot(data=df, x='Outcome', y='Glucose', hue='Outcome', split=False, palette='Set1')
plt.title('Glucose Distribution (Violin)')
plt.show()
```

```python
# stripplot: shows every individual data point per category
plt.figure(figsize=(7, 5))
sns.stripplot(data=df, x='Outcome', y='Glucose', jitter=True, alpha=0.5, palette='Set1')
# jitter adds tiny random horizontal spread so points don't stack
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Glucose Stripplot by Outcome')
plt.show()
```

```python
# swarmplot: like stripplot but no overlap (points arranged neatly)
plt.figure(figsize=(7, 5))
sns.swarmplot(data=df.sample(100), x='Outcome', y='Glucose', palette='coolwarm')
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Glucose Swarmplot (100 samples)')
plt.show()
```

```python
# barplot: shows mean + confidence interval per category
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x='Outcome', y='Glucose', palette='Blues', ci=95)
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Mean Glucose by Outcome (Barplot)')
plt.show()
```

```python
# countplot: count of rows per category (like a bar chart of frequencies)
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Outcome', palette='pastel')
plt.xticks([0, 1], ['No Diabetes', 'Diabetes'])
plt.title('Count of Each Outcome')
plt.show()
```

```python
# Add Age groups column for catplot demo
df['Age_Group'] = pd.cut(df['Age'], bins=3, labels=['Young', 'Middle', 'Senior'])

# catplot: unified interface for all categorical plots — set kind= to switch
sns.catplot(data=df, x='Age_Group', y='Glucose', kind='box', hue='Outcome',
            palette='Set2', height=5, aspect=1.5)
plt.title('Glucose by Age Group and Outcome')
plt.show()
```

```python
# factorplot is the old name for catplot (same thing)
# catplot with kind='violin'
sns.catplot(data=df, x='Outcome', y='BMI', kind='violin',
            palette='muted', height=5, aspect=1.2)
plt.title('BMI Violin via catplot')
plt.show()
```

## 10. Matrix Plots — Heatmap

```python
# Heatmap of correlation matrix
plt.figure(figsize=(10, 7))
corr = df.drop('Age_Group', axis=1).corr()    # compute correlation
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
# annot=True prints the number in each cell
# fmt='.2f' formats numbers to 2 decimal places
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Heatmap with different colormap
plt.figure(figsize=(10, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='YlGnBu', linewidths=0.5, linecolor='white')
plt.title('Correlation Heatmap (YlGnBu)')
plt.show()
```

```python
# Heatmap: mean values of features grouped by Outcome
pivot = df.drop('Age_Group', axis=1).groupby('Outcome').mean()
plt.figure(figsize=(10, 3))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='RdYlGn', linewidths=0.5)
plt.title('Mean Feature Values by Outcome')
plt.yticks([0.5, 1.5], ['No Diabetes', 'Diabetes'], rotation=0)
plt.show()
```

```python
# Clustermap: heatmap + hierarchical clustering (groups similar columns/rows)
cluster_data = df.drop(['Outcome', 'Age_Group'], axis=1).sample(50)
sns.clustermap(cluster_data.corr(), annot=True, fmt='.2f', cmap='coolwarm', figsize=(9, 7))
plt.title('Clustermap of Feature Correlations')
plt.show()
```
