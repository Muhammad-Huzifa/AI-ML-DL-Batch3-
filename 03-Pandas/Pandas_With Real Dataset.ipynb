## 1.1 What is Pandas?

**Pandas** is the most popular Python library for **data manipulation and analysis**.

| Feature | Description |
|---------|-------------|
| Full Name | **Pan**el **Da**ta |
| Created By | Wes McKinney (2008) |
| Built On | NumPy |
| Primary Use | Data cleaning, exploration, preprocessing |
| In ML/DL | Bridges raw data ➡️ ML-ready arrays |

### Why Pandas in Machine Learning?
```
Raw CSV/Excel  ──►  Pandas (clean, transform, explore)  ──►  NumPy Array  ──►  ML Model
```
Before ANY model training, data must be:
- Loaded from files
- Cleaned (remove NaN, fix wrong values)
- Transformed (encoding, scaling)
- Explored (statistics, distributions)

**All of this is done with Pandas.**

## 1.2 Two Core Data Structures

Pandas has exactly **2 main data structures** you must master:

| Structure | Dimensions | Think of it as... |
|-----------|------------|--------------------|
| **Series** | 1D | A single column of a spreadsheet |
| **DataFrame** | 2D | A full spreadsheet / SQL table |

A **DataFrame** is simply a collection of **Series** sharing the same index.

````python
# ============================================================
# CELL 1.2 — Creating a Series from scratch
# ============================================================

# A Series is a 1D labeled array — like ONE column of data
glucose_levels = pd.Series([148, 85, 183, 89, 137],
                            name='Glucose')
print("--- Basic Series ---")
print(glucose_levels)
print(f"\nType   : {type(glucose_levels)}")
print(f"dtype  : {glucose_levels.dtype}")
print(f"shape  : {glucose_levels.shape}")
````

````python
# ============================================================
# CELL 1.3 — Series with custom index (labels)
# ============================================================

# Instead of 0,1,2... you can give meaningful labels
glucose_named = pd.Series(
    [148, 85, 183, 89, 137],
    index=['Patient_A', 'Patient_B', 'Patient_C', 'Patient_D', 'Patient_E'],
    name='Glucose'
)
print("--- Series with custom index ---")
print(glucose_named)
print(f"\nAccess by label : {glucose_named['Patient_C']}")
print(f"Access by position: {glucose_named.iloc[2]}")
````

````python
# ============================================================
# CELL 1.4 — Creating a Series from a dictionary
# ============================================================

# Dictionary keys become the index, values become the data
bmi_dict = {'Patient_A': 33.6, 'Patient_B': 26.6,
            'Patient_C': 23.3, 'Patient_D': 28.1,
            'Patient_E': 43.1}

bmi_series = pd.Series(bmi_dict, name='BMI')
print("--- Series from Dictionary ---")
print(bmi_series)
````

````python
# ============================================================
# CELL 1.5 — Creating a DataFrame from scratch (dict of lists)
# ============================================================

# A DataFrame = multiple Series combined side by side
# Dictionary: keys = column names, values = list of data
data = {
    'Pregnancies': [6, 1, 8, 1, 0],
    'Glucose'    : [148, 85, 183, 89, 137],
    'BloodPress' : [72, 66, 64, 66, 40],
    'BMI'        : [33.6, 26.6, 23.3, 28.1, 43.1],
    'Age'        : [50, 31, 32, 21, 33],
    'Outcome'    : [1, 0, 1, 0, 1]
}

df_manual = pd.DataFrame(data)
print("--- Manually Created DataFrame ---")
print(df_manual)
print(f"\nShape  : {df_manual.shape}   ← (rows, columns)")
print(f"Type   : {type(df_manual)}")
````

````python
# ============================================================
# CELL 1.6 — Creating DataFrame from list of dictionaries
# ============================================================

# Each dictionary = one row
records = [
    {'Name': 'Ali',   'Glucose': 148, 'BMI': 33.6, 'Outcome': 1},
    {'Name': 'Sara',  'Glucose': 85,  'BMI': 26.6, 'Outcome': 0},
    {'Name': 'Ahmed', 'Glucose': 183, 'BMI': 23.3, 'Outcome': 1},
]

df_records = pd.DataFrame(records)
print("--- DataFrame from list of dicts ---")
print(df_records)
````

````python
# ============================================================
# CELL 1.7 — Creating DataFrame from NumPy array
# ============================================================

# NumPy array → DataFrame (used when data already in array form)
arr = np.array([[6, 148, 72, 33.6, 50, 1],
                [1,  85, 66, 26.6, 31, 0],
                [8, 183, 64, 23.3, 32, 1]])

df_from_array = pd.DataFrame(
    arr,
    columns=['Pregnancies','Glucose','BloodPress','BMI','Age','Outcome']
)
print("--- DataFrame from NumPy Array ---")
print(df_from_array)
````

## 2.3 Row Selection — loc and iloc

````python
# ============================================================
# CELL 2.8 — loc[] → Label-Based Selection
# ============================================================
#
#  df.loc[row_label, column_label]
#  ▸ Uses NAMES/LABELS of rows and columns
#  ▸ END of slice is INCLUSIVE
#  ▸ Works with boolean arrays too
# ============================================================

# Select a single row by label
print("--- Single row by label ---")
print(df.loc[0])
print()
````

````python
# loc with row range — END IS INCLUSIVE (0,1,2,3,4 = 5 rows)
print("--- Rows 0 to 4 (inclusive) ---")
print(df.loc[0:4])
````

````python
# loc with row AND column labels
print("--- Specific rows AND columns ---")
print(df.loc[0:4, ['Glucose', 'BMI', 'Outcome']])
````

````python
# loc: single cell — one row, one column
print("--- Single cell value ---")
value = df.loc[5, 'Glucose']
print(f"Glucose of patient at row 5: {value}")
````

````python
# loc: column range (from Glucose to BMI)
print("--- Column range using loc ---")
print(df.loc[0:3, 'Glucose':'BMI'])
````

````python
# ============================================================
# CELL 2.9 — iloc[] → Position-Based Selection
# ============================================================
#
#  df.iloc[row_position, col_position]
#  ▸ Uses INTEGER POSITIONS (0, 1, 2...)
#  ▸ END of slice is EXCLUSIVE (like Python lists)
#  ▸ Does NOT know column names — only positions
# ============================================================

# Select first row by position
print("--- First row by position ---")
print(df.iloc[0])
````

````python
# iloc: rows 0,1,2 (NOT 3) — end is EXCLUSIVE
print("--- Rows 0,1,2 (exclusive end) ---")
print(df.iloc[0:3])
````

````python
# iloc: rows AND columns by position
print("--- First 4 rows, first 3 columns ---")
print(df.iloc[0:4, 0:3])
````

````python
# iloc: single cell by position
print("--- Single cell: row 0, col 1 (Glucose) ---")
print(df.iloc[0, 1])
````

````python
# iloc: negative indexing — last row
print("--- Last row ---")
print(df.iloc[-1])
````

````python
# iloc: last 5 rows
print("--- Last 5 rows ---")
print(df.iloc[-5:])
````

````python
# iloc: all rows, specific columns by position
print("--- All rows, columns at position 1, 4, 8 ---")
print(df.iloc[:, [1, 4, 8]].head())
````

````python
# ============================================================
# CELL 2.10 — loc vs iloc — KEY DIFFERENCE DEMO
# ============================================================

# Create a DataFrame with a NON-default index to show the difference clearly
df_test = pd.DataFrame(
    {'Glucose': [148, 85, 183, 89, 137],
     'Outcome': [1, 0, 1, 0, 1]},
    index=[10, 20, 30, 40, 50]  # index labels are 10,20,30,40,50
)

print("DataFrame (index = 10,20,30,40,50):")
print(df_test)
print()

# loc uses the LABEL 30
print("df.loc[30]  → row with LABEL 30:")
print(df_test.loc[30])
print()

# iloc uses POSITION 2 (the 3rd row)
print("df.iloc[2]  → row at POSITION 2 (also the row with label 30):")
print(df_test.iloc[2])
````

## 2.4 Conditional Selection (Boolean Filtering)

````python
# ============================================================
# CELL 2.11 — Basic boolean / conditional selection
# ============================================================

# Step 1: A condition creates a boolean Series (True/False per row)
condition = df['Glucose'] > 140
print("Boolean Series (first 10):")
print(condition.head(10))
print(f"\n{condition.sum()} rows have Glucose > 140")
````

````python
# Step 2: Use the boolean Series to filter the DataFrame
high_glucose = df[df['Glucose'] > 140]
print("Patients with Glucose > 140:")
print(high_glucose[['Glucose','BMI','Age','Outcome']].head(8))
print(f"\nTotal: {len(high_glucose)} patients")
````

````python
# ============================================================
# CELL 2.12 — AND condition (&) — both conditions must be True
# ============================================================

# IMPORTANT: Use & not 'and', and wrap each condition in ()
risky = df[(df['Glucose'] > 140) & (df['BMI'] > 30)]
print("High Glucose AND High BMI patients:")
print(risky[['Glucose','BMI','Age','Outcome']].head())
print(f"\nTotal: {len(risky)} patients")
````

````python
# ============================================================
# CELL 2.13 — OR condition (|) — at least one must be True
# ============================================================

at_risk = df[(df['Age'] > 50) | (df['Glucose'] > 160)]
print("Age > 50 OR Glucose > 160:")
print(at_risk[['Age','Glucose','BMI','Outcome']].head())
print(f"\nTotal: {len(at_risk)} patients")
````

````python
# ============================================================
# CELL 2.14 — NOT condition (~) — negates the condition
# ============================================================

# ~ flips True→False and False→True
non_diabetic = df[~(df['Outcome'] == 1)]   # same as df[df['Outcome'] == 0]
print(f"Non-diabetic patients: {len(non_diabetic)}")
print(non_diabetic[['Glucose','BMI','Age','Outcome']].head())
````

````python
# ============================================================
# CELL 2.15 — isin() — check membership in a list
# ============================================================

# Select patients who are exactly age 21, 25, or 30
specific_ages = df[df['Age'].isin([21, 25, 30])]
print("Patients aged 21, 25, or 30:")
print(specific_ages[['Age','Glucose','Outcome']].head(10))
print(f"\nTotal: {len(specific_ages)}")
````

````python
# ============================================================
# CELL 2.16 — between() — range filtering (inclusive both ends)
# ============================================================

middle_aged = df[df['Age'].between(30, 50)]
print(f"Patients aged 30-50: {len(middle_aged)}")
print(middle_aged['Age'].describe())
````

````python
# ============================================================
# CELL 2.17 — query() — SQL-style string filtering
# ============================================================

# query() lets you write conditions as readable strings
result = df.query('Glucose > 140 and Age > 40 and Outcome == 1')
print("query(): Glucose>140 AND Age>40 AND Diabetic")
print(result[['Glucose','Age','BMI','Outcome']].head())
print(f"Total: {len(result)}")
````

## 3.4 Applying Functions — apply(), map()

````python
# ============================================================
# CELL 3.6 — apply() with a named function on a Series
# ============================================================

def categorize_bmi(bmi):
    """Classify BMI into standard categories"""
    if bmi < 18.5:  return 'Underweight'
    elif bmi < 25:  return 'Normal'
    elif bmi < 30:  return 'Overweight'
    else:           return 'Obese'

# Apply the function to every value in the BMI column
df['BMI_Category'] = df['BMI'].apply(categorize_bmi)
print("BMI Category distribution:")
print(df['BMI_Category'].value_counts())
````

````python
# ============================================================
# CELL 3.7 — apply() with lambda (one-liner function)
# ============================================================

# Lambda = anonymous function written in one line
# lambda x: expression   ← x is the input (one cell value)

# Scale Glucose to 0-1 range
df['Glucose_Scaled'] = df['Glucose'].apply(
    lambda x: (x - df['Glucose'].min()) / (df['Glucose'].max() - df['Glucose'].min())
)
print("Original vs Scaled Glucose:")
print(df[['Glucose', 'Glucose_Scaled']].head())
print(f"\nMin: {df['Glucose_Scaled'].min():.4f}  Max: {df['Glucose_Scaled'].max():.4f}")
````

````python
# ============================================================
# CELL 3.8 — apply() on rows (axis=1) — uses multiple columns
# ============================================================

# axis=0 = apply function column-by-column (default)
# axis=1 = apply function row-by-row

def risk_label(row):
    """Use multiple columns to compute a custom risk label per patient"""
    score = 0
    if row['Glucose'] > 140:       score += 2
    if row['BMI'] > 30:            score += 1
    if row['Age'] > 45:            score += 1
    if row['Pregnancies'] > 5:     score += 1

    if score >= 4:   return 'High Risk'
    elif score >= 2: return 'Medium Risk'
    else:            return 'Low Risk'

# axis=1 passes an entire ROW as the argument each time
df['Risk_Label'] = df.apply(risk_label, axis=1)
print("Risk Label distribution:")
print(df['Risk_Label'].value_counts())
````

````python
# ============================================================
# CELL 3.9 — apply() on entire DataFrame column-wise
# ============================================================

# Apply np.mean to every numeric column at once
numeric_cols = ['Pregnancies','Glucose','BloodPressure','SkinThickness',
                'Insulin','BMI','DiabetesPedigreeFunction','Age']

means = df[numeric_cols].apply(np.mean)
print("Column means via apply(np.mean):")
print(means)
````

````python
# ============================================================
# CELL 3.10 — map() — element-wise mapping using a dictionary
# ============================================================

# map() works on a Series only (not DataFrame)
# Perfect for replacing values with a lookup dictionary

df['Outcome_Text'] = df['Outcome'].map({0: 'No Diabetes', 1: 'Diabetes'})
print("Mapped Outcome:")
print(df[['Outcome','Outcome_Text']].head(10))
````

````python
# map() with a function (same as apply for Series)
df['Glucose_Round'] = df['Glucose'].map(lambda x: round(x, -1))  # round to nearest 10
print("Glucose rounded to nearest 10:")
print(df[['Glucose','Glucose_Round']].head())
````

````python
# ============================================================
# CELL 3.15 — Identify impossible zero values (domain knowledge)
# ============================================================

# In THIS dataset, 0 in medical columns = impossible = actual missing
zero_cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']

print("Zero counts per medical column (= hidden missing values):")
for col in zero_cols:
    zeros = (df[col] == 0).sum()
    pct   = zeros / len(df) * 100
    print(f"  {col:28s}: {zeros:4d} zeros  ({pct:.1f}%)")
````

## 3.7 Value Replacement

````python
# ============================================================
# CELL 3.16 — replace() — replace specific values
# ============================================================

df_clean = df.copy()   # always work on a copy!

# Replace 0 with NaN in medical columns — single column
df_clean['Glucose'] = df_clean['Glucose'].replace(0, np.nan)
print(f"Glucose zeros after replace: {(df_clean['Glucose'] == 0).sum()}")
print(f"Glucose NaNs  after replace: {df_clean['Glucose'].isna().sum()}")
````

````python
# Replace 0 with NaN across multiple columns at once
df_clean[zero_cols] = df_clean[zero_cols].replace(0, np.nan)
print("Missing values after replacing 0s:")
print(df_clean.isna().sum())
````

````python
# replace() with a dictionary — multiple replacements at once
df_temp = df[['Outcome']].copy()
df_temp['Outcome'] = df_temp['Outcome'].replace({0: 'Negative', 1: 'Positive'})
print("Replaced 0/1 with labels:")
print(df_temp['Outcome'].value_counts())
````

````python
# replace() with regex=True — pattern-based replacement
df_str = pd.DataFrame({'col': ['  good  ', '  bad  ', '  neutral  ']})
df_str['col_clean'] = df_str['col'].str.strip()  # remove spaces
print("String cleaning with str.strip():")
print(df_str)
````

## 3.8 Dropping Rows and Columns

````python
# ============================================================
# CELL 3.17 — drop() columns
# ============================================================

# Drop a single column
df_temp = df.copy()
df_temp = df_temp.drop(columns=['Outcome_Text'])
print("After dropping Outcome_Text:", df_temp.shape)

# Drop multiple columns
df_temp2 = df.copy()
df_temp2 = df_temp2.drop(columns=['BMI_Category','Glucose_Scaled',
                                   'Risk_Label','Outcome_Text',
                                   'Glucose_Round','Patient_ID'])
print("After dropping helper columns:", df_temp2.columns.tolist())
````

````python
# ============================================================
# CELL 3.18 — drop() rows by index label
# ============================================================

df_temp = df.head(10).copy()
print("Before drop — index:", df_temp.index.tolist())

# Drop rows with label 0, 3, 7
df_temp = df_temp.drop(index=[0, 3, 7])
print("After drop([0,3,7]) — index:", df_temp.index.tolist())
````

````python
# ============================================================
# CELL 3.19 — inplace=True vs reassignment
# ============================================================

df_a = df[['Glucose','BMI','Age','Outcome']].head(5).copy()

# Method 1: inplace=True — modifies original directly
df_a.drop(columns=['Age'], inplace=True)
print("inplace=True:", df_a.columns.tolist())

# Method 2: reassignment (safer — you keep original)
df_b = df[['Glucose','BMI','Age','Outcome']].head(5).copy()
df_b = df_b.drop(columns=['BMI'])
print("Reassignment:", df_b.columns.tolist())
````

````python
# ============================================================
# CELL 3.20 — Adding new columns
# ============================================================

df_work = df[['Pregnancies','Glucose','BMI','Age','Outcome']].copy()

# Simple arithmetic — add two columns
df_work['Glucose_plus_Age'] = df_work['Glucose'] + df_work['Age']

# NumPy operation on a column
df_work['Log_Glucose'] = np.log1p(df_work['Glucose'])  # log(1+x) handles 0 safely

# Condition-based column with np.where
df_work['Senior'] = np.where(df_work['Age'] >= 50, 1, 0)

print(df_work.head())
print("\nSenior patients:", df_work['Senior'].sum())
````

````python
# ============================================================
# CELL 3.21 — insert() — add column at specific position
# ============================================================

df_work2 = df[['Glucose','BMI','Age','Outcome']].copy()

# insert(position, column_name, values)
df_work2.insert(1, 'Glucose_Category', np.where(df_work2['Glucose']>125, 'High','Normal'))

print("Column inserted at position 1:")
print(df_work2.head())
````

````python
# ============================================================
# CELL 3.22 — duplicated() and drop_duplicates()
# ============================================================

print("Total duplicate rows:", df.duplicated().sum())

# Drop duplicates — keep first occurrence
df_no_dup = df.drop_duplicates()
print(f"Before: {len(df)}  After drop_duplicates: {len(df_no_dup)}")

# Check duplicates based on specific columns only
dup_subset = df.duplicated(subset=['Glucose','BMI','Age']).sum()
print(f"Rows with same Glucose+BMI+Age: {dup_subset}")
````

````python
# ============================================================
# CELL 3.23 — astype() — convert column data types
# ============================================================

print("Before astype:")
print(df[['Pregnancies','Glucose','Outcome']].dtypes)

df_typed = df[['Pregnancies','Glucose','Outcome']].copy()
df_typed['Pregnancies'] = df_typed['Pregnancies'].astype(np.int32)   # smaller int
df_typed['Glucose']     = df_typed['Glucose'].astype(np.float32)     # smaller float
df_typed['Outcome']     = df_typed['Outcome'].astype(bool)           # boolean

print("\nAfter astype:")
print(df_typed.dtypes)
````

````python
# ============================================================
# CELL 3.24 — Statistical operations on columns/rows
# ============================================================

num_cols = ['Pregnancies','Glucose','BloodPressure','BMI','Age']

print("Column means  :"); print(df[num_cols].mean().round(2))
print("\nColumn medians:"); print(df[num_cols].median().round(2))
print("\nColumn std    :"); print(df[num_cols].std().round(2))
print("\nColumn min    :"); print(df[num_cols].min())
print("\nColumn max    :"); print(df[num_cols].max())
````

## 4.1 Types of Missing Data

| Type | Full Name | Meaning | Example |
|------|-----------|---------|--------|
| **MCAR** | Missing Completely At Random | No pattern — random | Sensor glitch |
| **MAR** | Missing At Random | Depends on other observed data | Older patients skip some tests |
| **MNAR** | Missing Not At Random | Depends on the missing value itself | Very sick patients skip checkups |

In our diabetes dataset: **zeros in Glucose/Insulin/BMI are MNAR** — the missing values are related to the medical condition itself.

````python
# ============================================================
# CELL 4.1 — Reload clean version for missing data analysis
# ============================================================

df_raw = pd.read_csv(r'C:\Users\Mhuza\Downloads\archive (1).zip\diabetes.csv')
print("Fresh dataset loaded:", df_raw.shape)
df_raw.head()
````

````python
# ============================================================
# CELL 4.2 — Detect NaN with isna() / isnull()
# ============================================================

# isna() and isnull() are identical — both check for NaN/None
print("=== Missing value matrix (first 5 rows) ===")
print(df_raw.isna().head())
print()
print("=== Missing count per column ===")
print(df_raw.isna().sum())
````

````python
# ============================================================
# CELL 4.3 — Introduce real missing values (replace zeros)
# ============================================================

df_miss = df_raw.copy()
zero_medical_cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']

# Replace biologically impossible zeros with NaN
df_miss[zero_medical_cols] = df_miss[zero_medical_cols].replace(0, np.nan)

print("=== Missing values after replacing zeros ===")
missing_report = pd.DataFrame({
    'Missing Count': df_miss.isna().sum(),
    'Missing %':     (df_miss.isna().sum() / len(df_miss) * 100).round(1)
})
print(missing_report[missing_report['Missing Count'] > 0])
````

````python
# ============================================================
# CELL 4.4 — Visualize missing data
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart of missing counts
missing_counts = df_miss.isna().sum().sort_values(ascending=False)
missing_counts[missing_counts > 0].plot(
    kind='bar', ax=axes[0], color='salmon', edgecolor='black'
)
axes[0].set_title('Missing Value Counts per Column', fontsize=13)
axes[0].set_ylabel('Number of Missing Values')
axes[0].tick_params(axis='x', rotation=45)

# Heatmap of missing data locations
sns.heatmap(df_miss[zero_medical_cols].isna(),
            cbar=False, yticklabels=False,
            cmap='Reds', ax=axes[1])
axes[1].set_title('Missing Data Locations (red = missing)', fontsize=13)

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# CELL 4.5 — notna() — find rows WITHOUT missing values
# ============================================================

# notna() = opposite of isna()
complete_rows = df_miss[df_miss['Insulin'].notna()]
print(f"Rows WITH insulin data    : {len(complete_rows)}")
print(f"Rows WITHOUT insulin data : {df_miss['Insulin'].isna().sum()}")
````

## 4.2 Strategy 1 — dropna() — Remove Missing Data

````python
# ============================================================
# CELL 4.6 — dropna() — drop rows with ANY missing value
# ============================================================

df_dropped_any = df_miss.dropna()
print(f"Original : {len(df_miss)} rows")
print(f"After dropna()     : {len(df_dropped_any)} rows")
print(f"Rows removed       : {len(df_miss) - len(df_dropped_any)}")
print(f"Data loss          : {(1 - len(df_dropped_any)/len(df_miss))*100:.1f}%")
print("\n⚠️  Dropping 48% of data is too much — imputation is better here!")
````

````python
# ============================================================
# CELL 4.7 — dropna(subset=[]) — drop only if SPECIFIC columns are NaN
# ============================================================

# Only drop rows where GLUCOSE is missing (most important feature)
df_drop_glucose = df_miss.dropna(subset=['Glucose'])
print(f"Rows after dropping missing Glucose: {len(df_drop_glucose)}")
print(f"Still missing in Insulin: {df_drop_glucose['Insulin'].isna().sum()}")
````

````python
# ============================================================
# CELL 4.8 — dropna(thresh=n) — keep rows with at least n non-NaN
# ============================================================

# Keep rows that have at least 7 non-null values (out of 9 columns)
df_thresh = df_miss.dropna(thresh=7)
print(f"Rows with at least 7 non-null values: {len(df_thresh)}")
print(f"Rows dropped: {len(df_miss) - len(df_thresh)}")
````

````python
# ============================================================
# CELL 4.9 — dropna(axis=1) — drop COLUMNS with missing values
# ============================================================

df_col_drop = df_miss.dropna(axis=1)
print(f"Original columns  : {df_miss.columns.tolist()}")
print(f"After axis=1 drop : {df_col_drop.columns.tolist()}")
print("⚠️  Columns with ANY NaN are completely removed!")
````

## 4.3 Strategy 2 — fillna() — Impute Missing Data

````python
# ============================================================
# CELL 4.10 — fillna() with a CONSTANT value
# ============================================================

df_fill = df_miss.copy()

# Fill with zero (NOT recommended for medical data)
df_fill_zero = df_fill.copy()
df_fill_zero['Insulin'] = df_fill_zero['Insulin'].fillna(0)
print(f"After fill with 0 — Insulin NaN: {df_fill_zero['Insulin'].isna().sum()}")
````

````python
# ============================================================
# CELL 4.11 — fillna() with MEAN — simple imputation
# ============================================================

df_mean = df_miss.copy()

glucose_mean = df_mean['Glucose'].mean()
df_mean['Glucose'] = df_mean['Glucose'].fillna(glucose_mean)

print(f"Glucose mean used for imputation: {glucose_mean:.2f}")
print(f"Glucose NaN after fill: {df_mean['Glucose'].isna().sum()}")
````

````python
# ============================================================
# CELL 4.12 — fillna() with MEDIAN — better for skewed data
# ============================================================

# Median is robust to outliers — USE THIS for medical data
df_median = df_miss.copy()

for col in zero_medical_cols:
    median_val = df_median[col].median()
    df_median[col] = df_median[col].fillna(median_val)
    print(f"{col:20s} → median = {median_val:.2f}  | NaN after fill: {df_median[col].isna().sum()}")

print(f"\n✅ Total NaN remaining: {df_median.isna().sum().sum()}")
````

````python
# ============================================================
# CELL 4.13 — fillna() with MODE — best for categorical columns
# ============================================================

# Mode = most frequently occurring value
# Good for categorical features with missing values
demo = pd.Series(['A', 'B', 'A', np.nan, 'B', 'A', np.nan])
mode_val = demo.mode()[0]   # mode() returns a Series — take index 0
demo_filled = demo.fillna(mode_val)
print(f"Original : {demo.tolist()}")
print(f"Mode     : {mode_val}")
print(f"Filled   : {demo_filled.tolist()}")
````

````python
# ============================================================
# CELL 4.14 — fillna() with ffill — Forward Fill
# ============================================================

# Forward fill: fill NaN with the PREVIOUS non-NaN value
# Very useful for time-series data (e.g., stock prices)
ts_data = pd.Series([100, np.nan, np.nan, 103, np.nan, 105])
print("Original    :", ts_data.tolist())
print("ffill       :", ts_data.ffill().tolist())
````

````python
# ============================================================
# CELL 4.15 — fillna() with bfill — Backward Fill
# ============================================================

# Backward fill: fill NaN with the NEXT non-NaN value
print("Original    :", ts_data.tolist())
print("bfill       :", ts_data.bfill().tolist())
````

````python
# ============================================================
# CELL 4.16 — Group-based imputation (ADVANCED, POWERFUL)
# ============================================================

# Fill Glucose with the MEDIAN of that patient's Outcome group
# (diabetic patients have different typical glucose than non-diabetic)

df_group = df_miss.copy()

# groupby + transform: compute group median, then use as fill value
group_medians = df_group.groupby('Outcome')['Glucose'].transform('median')
df_group['Glucose'] = df_group['Glucose'].fillna(group_medians)

print("Group-based median fill for Glucose:")
print(df_group.groupby('Outcome')['Glucose'].median())
print(f"\nGlucose NaN after group fill: {df_group['Glucose'].isna().sum()}")
````

## 4.4 Strategy 3 — Sklearn Imputers (ML-Grade)

````python
# ============================================================
# CELL 4.17 — SimpleImputer (sklearn)
# ============================================================

from sklearn.impute import SimpleImputer

df_ski = df_miss[zero_medical_cols].copy()

# strategy = 'mean', 'median', 'most_frequent', or 'constant'
imputer = SimpleImputer(strategy='median')
df_imputed = pd.DataFrame(
    imputer.fit_transform(df_ski),
    columns=zero_medical_cols
)

print("After SimpleImputer(strategy='median'):")
print(df_imputed.isna().sum())
print("\nImputed medians used:")
for col, val in zip(zero_medical_cols, imputer.statistics_):
    print(f"  {col}: {val:.2f}")
````

````python
# ============================================================
# CELL 4.18 — KNNImputer — impute using K nearest neighbors
# ============================================================

from sklearn.impute import KNNImputer

# KNN finds the k most similar patients and uses their average
# Much smarter than simple mean/median!

df_knn_input = df_miss[zero_medical_cols].copy()

knn_imputer = KNNImputer(n_neighbors=5)  # use 5 nearest neighbors
df_knn = pd.DataFrame(
    knn_imputer.fit_transform(df_knn_input),
    columns=zero_medical_cols
)

print("After KNNImputer (n_neighbors=5):")
print(df_knn.isna().sum())
print("\nSample of KNN-imputed Insulin values:")
print(df_knn['Insulin'].head(10).round(1).tolist())
````

## 4.5 Compare Imputation Strategies Visually

````python
# ============================================================
# CELL 4.19 — Compare distribution before vs after imputation
# ============================================================

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
col_to_check = 'Insulin'

# Original (with NaN dropped for plotting)
df_miss[col_to_check].dropna().plot(
    kind='hist', bins=30, ax=axes[0], color='steelblue',
    title=f'{col_to_check} — Original (no NaN)', edgecolor='black'
)

# Mean imputed
mean_fill = df_miss[col_to_check].fillna(df_miss[col_to_check].mean())
mean_fill.plot(
    kind='hist', bins=30, ax=axes[1], color='salmon',
    title=f'{col_to_check} — Mean Imputed', edgecolor='black'
)

# Median imputed
median_fill = df_miss[col_to_check].fillna(df_miss[col_to_check].median())
median_fill.plot(
    kind='hist', bins=30, ax=axes[2], color='mediumseagreen',
    title=f'{col_to_check} — Median Imputed', edgecolor='black'
)

for ax in axes:
    ax.set_xlabel(col_to_check)
    ax.set_ylabel('Frequency')

plt.suptitle('Effect of Imputation Strategy on Distribution', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
````

## 4.6 Final Complete Preprocessing Pipeline

````python
# ============================================================
# CELL 4.20 — COMPLETE CLEAN PIPELINE (put it all together)
# ============================================================

print("="*55)
print(" COMPLETE PANDAS PREPROCESSING PIPELINE")
print("="*55)

# ── STEP 1: Load ──────────────────────────────────────────
df_final = pd.read_csv(r'C:\Users\Mhuza\Downloads\archive (1).zip\diabetes.csv')
print(f"\n[1] Loaded        → {df_final.shape[0]} rows, {df_final.shape[1]} cols")

# ── STEP 2: Replace impossible zeros ──────────────────────
medical = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
df_final[medical] = df_final[medical].replace(0, np.nan)
total_missing = df_final.isna().sum().sum()
print(f"[2] Zeros→NaN     → {total_missing} missing cells found")

# ── STEP 3: Drop duplicates ────────────────────────────────
before = len(df_final)
df_final.drop_duplicates(inplace=True)
print(f"[3] Duplicates    → {before - len(df_final)} rows removed")

# ── STEP 4: Impute missing values with median ──────────────
for col in medical:
    df_final[col] = df_final[col].fillna(df_final[col].median())
print(f"[4] Imputed       → {df_final.isna().sum().sum()} NaN remaining")

# ── STEP 5: Reset index ────────────────────────────────────
df_final.reset_index(drop=True, inplace=True)
print(f"[5] Index reset   → 0 to {len(df_final)-1}")

# ── STEP 6: Final check ────────────────────────────────────
print(f"[6] Final shape   → {df_final.shape}")
print(f"[7] Any NaN left? → {df_final.isna().any().any()}")

print("\n" + "="*55)
print(" DATASET IS CLEAN AND ML-READY!")
print("="*55)
print()
df_final.head()
````
