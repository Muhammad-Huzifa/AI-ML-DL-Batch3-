```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
np.random.seed(42)

# Study hours between 1 and 10
X = np.random.uniform(1, 10, 100).reshape(-1, 1)

# Pass if hours > 5
y = (X.flatten() > 5).astype(int)

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)



model = LogisticRegression()
model.fit(X, y)

predictions = model.predict(X)
accuracy = accuracy_score(y, predictions)
print(f"Model Accuracy: {accuracy:.2f}")
# Create smooth range for plotting
X_test = np.linspace(1, 10, 200).reshape(-1, 1)

# Predict probabilities
y_prob = model.predict_proba(X_test)[:, 1]

# Plot
plt.scatter(X, y, label="Actual Data")
plt.plot(X_test, y_prob, label="Predicted Probability")

# Decision boundary (where probability = 0.5)
decision_boundary = model.intercept_[0] * -1 / model.coef_[0][0]
plt.axvline(decision_boundary, linestyle='--', label="Decision Boundary")

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression Decision Boundary")
plt.legend()
plt.show()


new_hours = np.array([[3], [6], [9]])
y1 = (new_hours.flatten() > 5).astype(int)

# Predict class (0 = Fail, 1 = Pass)
new_predictions = model.predict(new_hours)

print("Predictions:", new_predictions)
accuracy1 = accuracy_score(y1, new_predictions)
print(f"Model Accuracy: {accuracy1:.2f}")
```
